## 1. Sum (`np.sum`)

### Concept Explanation
`np.sum()` calculates the total sum of array elements along a specified axis or across the entire array.

### Real-world Example
Adding up the scores across all rounds in a multi-round sporting tournament to find a player's total score.

### Business Example
Calculating total gross revenue across all store branches for a given month.

### AI/ML Example
Summing pixel values or batch loss values during neural network backpropagation steps.

```python
import numpy as np

sales = np.array([
    [100, 150, 200],
    [110, 140, 190]
])

print("Total Sum (All elements):", np.sum(sales))
print("Sum along Columns (axis=0):", np.sum(sales, axis=0))
print("Sum along Rows (axis=1):", np.sum(sales, axis=1))
```

## 2. Mean (`np.mean`)

### Concept Explanation
`np.mean()` computes the arithmetic average of elements by taking the sum and dividing by the count.

### Real-world Example
Calculating the average daily temperature for a city over a month.

### Business Example
Determining the average spending amount per transaction in an e-commerce platform.

### AI/ML Example
Calculating Mean Squared Error (MSE) or averaging gradients across a mini-batch during gradient descent.

```python
import numpy as np

scores = np.array([85, 90, 78, 92, 88])
print("Mean Score:", np.mean(scores))
```

## 3. Median (`np.median`)

### Concept Explanation
`np.median()` identifies the middle value in a sorted array, separating the upper half from the lower half. It is robust to extreme outliers.

### Real-world Example
Finding the median income of a city's population so extreme billionaire incomes don't skew the metric.

### Business Example
Analyzing median delivery times for packages to evaluate realistic logistics performance.

### AI/ML Example
Using median imputation to replace missing values in skewed numeric features during data preprocessing.

```python
import numpy as np

salaries = np.array([30000, 35000, 40000, 42000, 1000000]) # 1000000 is an outlier
print("Mean Salary (skewed):", np.mean(salaries))
print("Median Salary (robust):", np.median(salaries))
```

## 4. Variance (`np.var`)

### Concept Explanation
`np.var()` measures the average squared deviation of each data point from the mean, indicating how spread out the numbers are.

### Real-world Example
Comparing the consistency of two archers: both hit a 50-point average, but one has a tight cluster (low variance) while the other is scattered (high variance).

### Business Example
Measuring variation in product weight output on a factory manufacturing line to maintain quality control.

### AI/ML Example
Analyzing feature variance during Feature Selection (e.g., Variance Thresholding to drop zero-variance features).

```python
import numpy as np

data = np.array([10, 12, 14, 16, 18])
print("Variance:", np.var(data))
```

## 5. Standard Deviation (`np.std`)

### Concept Explanation
`np.std()` computes the square root of variance, expressing data dispersion in the exact same units as the original dataset.

### Real-world Example
Measuring the fluctuation range of daily stock prices around their average price.

### Business Example
Setting safety stock inventory levels based on demand volatility measured by standard deviation.

### AI/ML Example
Standardizing features using Z-score normalization: $(X - \mu) / \sigma$.

```python
import numpy as np

data = np.array([10, 12, 14, 16, 18])
print("Standard Deviation:", np.std(data))
```

## 6. Minimum & Maximum (`np.min`, `np.max`)

### Concept Explanation
`np.min()` and `np.max()` find the smallest and largest values in an array along a specific axis or overall.

### Real-world Example
Tracking the lowest and highest daily temperatures recorded during a winter week.

### Business Example
Identifying the highest sale price and lowest discount rate across a product line.

### AI/ML Example
Min-Max scaling to normalize features into a $[0, 1]$ range: $(X - X_{min}) / (X_{max} - X_{min})$.

```python
import numpy as np

prices = np.array([19.99, 5.49, 89.00, 12.50])
print("Minimum Price:", np.min(prices))
print("Maximum Price:", np.max(prices))
```

## 7. Argmin & Argmax (`np.argmin`, `np.argmax`)

### Concept Explanation
`np.argmin()` and `np.argmax()` return the **index position** of the minimum and maximum values rather than the actual values.

### Real-world Example
Determining which runner won a race by checking the index of the minimum finish time in an array.

### Business Example
Finding the specific month index that achieved the peak sales revenue.

### AI/ML Example
Extracting predicted class labels from multi-class classification model probability outputs (e.g., Softmax layer).

```python
import numpy as np

probabilities = np.array([0.05, 0.82, 0.13]) # Cat, Dog, Bird probabilities
predicted_class = np.argmax(probabilities)

print("Highest Probability Index (Predicted Class):", predicted_class)
```

## 8. Percentile (`np.percentile`)

### Concept Explanation
`np.percentile()` computes the $n$-th percentile of data, indicating the value below which a given percentage of observations falls.

### Real-world Example
Scoring in the 90th percentile on a standardized entrance exam means scoring higher than 90% of test takers.

### Business Example
Evaluating SLA performance by tracking 95th or 99th percentile server response latency times.

### AI/ML Example
Detecting extreme outliers by calculating the Interquartile Range (IQR = $Q_3 - Q_1$).

```python
import numpy as np

scores = np.array([50, 60, 70, 80, 90, 100])
print("50th Percentile (Median):", np.percentile(scores, 50))
print("90th Percentile:", np.percentile(scores, 90))
```

## 9. Dot Product (`np.dot`)

### Concept Explanation
`np.dot()` performs the algebraic dot product (inner product) of two vectors, summing the products of their corresponding elements.

### Real-world Example
Multiplying quantities of items bought by their individual unit prices to calculate the total bill amount.

### Business Example
Calculating weighted total performance scores for employees based on KPI weight vectors.

### AI/ML Example
Computing the linear combination of inputs and weights ($Z = W \cdot X + b$) in a neuron layer.

```python
import numpy as np

inputs = np.array([2.0, 3.0, 1.5])
weights = np.array([0.5, -0.2, 0.8])

dot_product = np.dot(inputs, weights)
print("Dot Product (W . X):", dot_product)
```

## 10. Matrix Multiplication (`np.matmul` or `@`)

### Concept Explanation
Matrix multiplication multiplies rows of the first matrix by columns of the second matrix. The inner dimensions must match $(m \times n) \cdot (n \times p) = (m \times p)$.

### Real-world Example
Calculating total ingredient requirements for multiple cookie recipes across various batch orders simultaneously.

### Business Example
Projecting total multi-region revenues for several product categories using price and volume matrices.

### AI/ML Example
Transforming feature representations across neural network dense layers ($H = XW$).

```python
import numpy as np

A = np.array([[1, 2], [3, 4]]) # Shape (2, 2)
B = np.array([[5, 6], [7, 8]]) # Shape (2, 2)

# Matrix multiplication using @ operator
result = A @ B
print("Matrix Multiplication Result (A @ B):\n", result)
```

## 11. Transpose (`np.transpose` or `.T`)

### Concept Explanation
Transposing flips a matrix over its diagonal, swapping its row and column indices. Shape $(m \times n)$ becomes $(n \times m)$.

### Real-world Example
Flipping a schedule table so that days become rows instead of columns.

### Business Example
Re-orienting tabular data structure to match expected import formatting for another system.

### AI/ML Example
Aligning matrix dimensions before performing dot products or matrix multiplications in weight updates.

```python
import numpy as np

matrix = np.array([[1, 2, 3], [4, 5, 6]]) # Shape (2, 3)
transposed = matrix.T                     # Shape (3, 2)

print("Original Matrix (2x3):\n", matrix)
print("Transposed Matrix (3x2):\n", transposed)
```

## 12. Basic Linear Algebra Operations (`np.linalg`)

### Concept Explanation
The `np.linalg` submodule provides linear algebra functions such as calculating matrix determinants (`np.linalg.det`), inverses (`np.linalg.inv`), and eigenvalues/eigenvectors (`np.linalg.eig`).

### Real-world Example
Determining if a system of equations has a unique solution by evaluating whether its coefficient determinant is non-zero.

### Business Example
Solving resource allocation optimization equations across complex supply chains.

### AI/ML Example
Principal Component Analysis (PCA) uses Eigenvalues and Eigenvectors to perform dimensionality reduction.

```python
import numpy as np

A = np.array([[4, 7], [2, 6]])

# Determinant
det_A = np.linalg.det(A)

# Matrix Inverse
inv_A = np.linalg.inv(A)

print("Determinant of A:", det_A)
print("Inverse of A:\n", inv_A)
```

##  5 Interview Questions with Answers

1. **What is the difference between element-wise multiplication (`*`) and matrix multiplication (`@`) in NumPy?**  
   * **Answer:** Element-wise multiplication (`*`) multiplies corresponding elements in two arrays of the same shape. Matrix multiplication (`@` or `np.matmul`) performs linear algebra dot multiplication between rows and columns.

2. **How do `axis=0` and `axis=1` work in aggregation functions like `np.sum`?**  
   * **Answer:** `axis=0` operates vertically across rows (collapsing rows to compute column totals). `axis=1` operates horizontally across columns (collapsing columns to compute row totals).

3. **Why is `np.median` preferred over `np.mean` for skewed distributions?**  
   * **Answer:** `np.mean` is sensitive to extreme outliers, which can distort the measure of central tendency. `np.median` relies on position ordering and remains unaffected by extreme tail values.

4. **What is the difference between `argmax` and `max`?**  
   * **Answer:** `np.max` returns the actual maximum numerical value within an array, whereas `np.argmax` returns the index location where that maximum value resides.

5. **When does a matrix NOT have an inverse?**  
   * **Answer:** A matrix cannot be inverted if it is singular, meaning its determinant is equal to 0 (`np.linalg.det(A) == 0`).
