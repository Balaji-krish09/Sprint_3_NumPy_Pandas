## 1. What is Pandas & Why Pandas?

### Concept Explanation
Pandas is an open-source Python library built on top of NumPy designed for high-performance processing, cleaning, transforming, and analyzing structured/tabular data using **Series** (1D) and **DataFrames** (2D).

### Real-world Example
Analyzing a spreadsheet of patient health metrics (age, blood pressure, heart rate) across multiple hospital visits.

### Business Example
Extracting, cleaning, and transforming monthly e-commerce transactions containing customer details, order amounts, and payment statuses.

### AI/ML Example
Preprocessing raw datasets, handling missing values, encoding categorical variables, and preparing feature matrices ($X$) and target vectors ($y$) for machine learning pipelines.

```python
import pandas as pd
import numpy as np

print(f"Pandas Version: {pd.__version__}")
```

## 2. Pandas Series

### Concept Explanation
A **Series** is a 1D labeled array capable of holding any data type. It consists of data values paired with an associated array of labels called the **Index**.

### Real-world Example
A list of daily temperature readings labeled with specific dates as indices (`2026-08-01`: 32.5°C).

### Business Example
A company's daily stock closing prices indexed by ticker symbols or trade dates.

### AI/ML Example
Extracting a single target variable column (e.g., `y_true` or `house_price`) from a tabular dataset.

```python
import pandas as pd

# Creating a Series from a List with custom index
temperatures = pd.Series([31.2, 32.5, 29.8, 30.1], index=['Mon', 'Tue', 'Wed', 'Thu'], name="Temp_C")

print("Pandas Series:")
print(temperatures)
print("\nValues:", temperatures.values)
print("Index:", temperatures.index)
```

## 3. Pandas DataFrame

### Concept Explanation
A **DataFrame** is a 2D labeled, size-mutable tabular data structure consisting of rows and columns, similar to a SQL table or Excel spreadsheet.

### Real-world Example
A school grade book containing student names, subject scores, attendance percentages, and final grades.

### Business Example
An enterprise HR database containing employee IDs, departments, salaries, hire dates, and performance ratings.

### AI/ML Example
The primary input dataset matrix ($X$) passed into ML frameworks like Scikit-Learn or XGBoost.

```python
import pandas as pd

# Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'Department': ['HR', 'Tech', 'Tech', 'Marketing'],
    'Salary': [50000, 75000, 85000, 60000]
}

df = pd.DataFrame(data)
print("Pandas DataFrame:")
print(df)
```

## 4. Creating DataFrames

### Concept Explanation
DataFrames can be constructed from various Python data structures, including dictionaries of lists, lists of dictionaries, 2D NumPy arrays, or Pandas Series objects.

### Real-world Example
Converting unstructured sensor output dictionaries collected during an IoT experiment into a structured table.

### Business Example
Converting JSON payloads received from a payment gateway API into a clean DataFrame for financial reporting.

### AI/ML Example
Converting simulated feature matrices generated via NumPy random distributions into labeled DataFrames for model experimentation.

```python
import pandas as pd
import numpy as np

# Method 1: From Dictionary of Lists
df_dict = pd.DataFrame({'Item': ['Laptop', 'Mouse'], 'Price': [1200, 25]})

# Method 2: From List of Dictionaries
df_list_dict = pd.DataFrame([{'User': 'A', 'Score': 90}, {'User': 'B', 'Score': 85}])

# Method 3: From 2D NumPy Array with defined column names
matrix = np.random.randint(10, 100, size=(3, 2))
df_numpy = pd.DataFrame(matrix, columns=['Metric_A', 'Metric_B'])

print("From 2D NumPy Array:\n", df_numpy)
```

## 5. Data Types in Pandas (`dtypes`)

### Concept Explanation
Pandas assigns data types (`int64`, `float64`, `object`, `bool`, `datetime64`, `category`) to each column. Inspecting and managing data types ensures memory efficiency and prevents type errors during mathematical operations.

### Real-world Example
Ensuring zip codes or telephone numbers are stored as strings (`object`) rather than integers to prevent stripping leading zeros.

### Business Example
Converting transaction timestamps from generic text (`object`) to `datetime64` to perform month-over-month trend analysis.

### AI/ML Example
Converting low-cardinality string columns into `category` types to reduce memory overhead and prepare for categorical encoding.

```python
import pandas as pd

data = {
    'ID': [101, 102, 103],
    'Product': ['Phone', 'Tablet', 'Monitor'],
    'Price': [699.99, 499.50, 199.00],
    'In_Stock': [True, False, True]
}

df = pd.DataFrame(data)

print("Data Types:\n", df.dtypes)
print("\nDataFrame Summary Info:")
df.info()
```

## 6. Indexing & Custom Index (`df.index`, `set_index`)

### Concept Explanation
The index serves as an axis label for DataFrame rows. By default, Pandas assigns an integer sequence (`0, 1, 2...`), but any unique column (like Employee ID or Timestamp) can be set as the row index.

### Real-world Example
Indexing climate data by exact dates (`YYYY-MM-DD`) to enable fast date-range querying.

### Business Example
Setting `Customer_ID` as the index to quickly look up account details without scanning the entire dataset.

### AI/ML Example
Setting explicit sample IDs or timestamps as indices to keep track of individual predictions during model evaluation.

```python
import pandas as pd

df = pd.DataFrame({
    'Emp_ID': ['E101', 'E102', 'E103'],
    'Name': ['Sarah', 'John', 'Elena'],
    'Role': ['Developer', 'Manager', 'Analyst']
})

# Setting Emp_ID as the index
df_indexed = df.set_index('Emp_ID')
print("DataFrame with Custom Index:\n", df_indexed)

# Resetting index back to default integer index
df_reset = df_indexed.reset_index()
print("\nDataFrame after Reset Index:\n", df_reset)
```

## 7. Working with Columns (`df.columns`, Renaming, Adding, Dropping)

### Concept Explanation
Columns represent attributes/variables in a tabular dataset. Pandas provides intuitive methods to access, rename, create derived columns, or drop unneeded features.

### Real-world Example
Adding a calculated `BMI` column using existing `Height` and `Weight` columns in health research.

### Business Example
Renaming column header names coming from an old SQL legacy database into clean, standardized formats.

### AI/ML Example
Engineering new features (e.g., `Total_Purchase = Quantity * Unit_Price`) or dropping redundant columns during Feature Selection.

```python
import pandas as pd

df = pd.DataFrame({
    'unit_price': [10, 20, 15],
    'qty': [3, 5, 2]
})

# 1. Access columns
print("Column Names:", df.columns.tolist())

# 2. Add derived column
df['total_cost'] = df['unit_price'] * df['qty']

# 3. Rename columns
df = df.rename(columns={'unit_price': 'UnitPrice', 'qty': 'Quantity', 'total_cost': 'TotalCost'})

# 4. Drop a column
df_reduced = df.drop(columns=['UnitPrice'])

print("\nModified DataFrame:\n", df)
print("\nAfter Dropping UnitPrice:\n", df_reduced)
```

## 9. Basic Writing Data (`df.to_csv`)

### Concept Explanation
`df.to_csv()` exports a Pandas DataFrame to a CSV file on local disk, with options to exclude the automatic index or specify custom delimiters.

### Real-world Example
Exporting cleaned observational survey results to share with research collaborators.

### Business Example
Saving aggregated monthly financial summaries to a CSV report for business intelligence tools like Power BI or Tableau.

### AI/ML Example
Saving model prediction results (`submission.csv`) containing ID and predicted target probabilities.

```python
import pandas as pd

df = pd.DataFrame({
    'ID': [101, 102],
    'Status': ['Approved', 'Pending']
})

# Exporting to CSV without writing the row index
df.to_csv("output_status.csv", index=False)
print("Successfully wrote DataFrame to 'output_status.csv'")
```

## Minimum 5 Interview Questions with Answers

1. **What is the structural difference between a Pandas Series and a Pandas DataFrame?**  
   * **Answer:** A Pandas Series is a 1-dimensional labeled array that holds a single data type, whereas a DataFrame is a 2-dimensional tabular structure with labeled rows and columns capable of storing heterogeneous data types.

2. **Why is Pandas built on top of NumPy, and how do they differ?**  
   * **Answer:** Pandas leverages NumPy's high-performance C-based array storage for numerical speed, but adds high-level tabular features like axis labels, handling mixed data types, missing value support, merging/joining, and string operations.

3. **What is the purpose of `df.set_index()` versus `df.reset_index()`?**  
   * **Answer:** `df.set_index()` converts an existing column into the DataFrame's row index. `df.reset_index()` restores the row index back to the default integer sequence (`0, 1, 2...`) and moves the index values back into a standard column.

4. **How do you optimize memory usage when loading a DataFrame with categorical string columns?**  
   * **Answer:** Convert high-frequency, low-cardinality string columns (e.g., Gender, State, Status) to the `category` data type using `df['col'] = df['col'].astype('category')`.

5. **Why should you usually pass `index=False` when using `df.to_csv()`?**  
   * **Answer:** By default, `to_csv()` exports the numeric Pandas row index (`0, 1, 2...`) as an extra unlabeled column. Setting `index=False` prevents creating unwanted extra columns when reloading the file.
