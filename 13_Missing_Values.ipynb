## 1. Detecting Missing Data (`isnull`, `notnull`)

### Concept Explanation
`isnull()` (an alias for `isna()`) returns a boolean mask marking `NaN`/`None` locations as `True`. `notnull()` (alias `notna()`) returns `True` for valid, non-missing values.

### Real-world Example
Auditing IoT weather station feeds to pinpoint offline sensor intervals.

### Business Example
Quantifying missing customer phone numbers prior to SMS marketing dispatch.

### AI/ML Example
Calculating missing value percentages per column to identify feature suitability during EDA.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1.0, 2.0, np.nan, 4.0, 5.0],
    'B': [np.nan, 20.0, 30.0, np.nan, 50.0],
    'C': ['X', 'Y', np.nan, 'Z', 'W']
})

print("Missing Values per Column:\n", df.isnull().sum())
print("\nValid Non-Null Rows in B:\n", df[df['B'].notnull()])
```

## 2. Dropping Missing Data (`dropna`)

### Concept Explanation
`dropna()` removes rows or columns containing missing values based on rules like `how='any'`, `how='all'`, or `subset=['col']`.

### Real-world Example
Discarding incomplete survey records missing required consent checkboxes.

### Business Example
Removing sales records missing transaction IDs before ledger reconciliation.

### AI/ML Example
Dropping training samples missing ground-truth target labels ($y$).

```python
import pandas as pd

# Drop rows missing values in column 'A'
df_clean_a = df.dropna(subset=['A'])

print("Shape after dropping rows where A is NaN:", df_clean_a.shape)
```

## 3. Imputing Missing Values (`fillna`, `interpolate`)

### Concept Explanation
* `fillna()` replaces `NaN` entries using constant values, calculated summary statistics (mean/median), or directional fills (`ffill`/`bfill`).
* `interpolate()` estimates missing numerical values by evaluating surrounding data points (e.g., linear estimation).

### Real-world Example
Estimating missing hourly temperature readings using linear interpolation across adjacent time bounds.

### Business Example
Replacing missing customer age values with dataset median age.

### AI/ML Example
Imputing missing feature values to preserve maximum training dataset volume.

```python
import pandas as pd

df_imp = df.copy()

# Linear interpolation for numerical series
df_imp['A_Interp'] = df_imp['A'].interpolate(method='linear')

# Median imputation
df_imp['B_Median'] = df_imp['B'].fillna(df_imp['B'].median())

print("Imputed Data:\n", df_imp[['A', 'A_Interp', 'B', 'B_Median']])
```

## 4. Missing Value Handling Strategies

### Concept Explanation
Different missing data scenarios require distinct strategies:
1. **Drop:** Best when missingness is under ~5% and strictly random.
2. **Mean/Median Imputation:** Best for skewed/normally distributed continuous variables missing at random.
3. **Mode/Category Imputation:** Best for categorical missing flags (e.g., replacing with `'Unknown'`).
4. **Interpolation:** Ideal for ordered time-series or sequential telemetry.

### Real-world Example
Choosing linear time interpolation for time-series climate sensors versus dropping empty user profile entries.

### Business Example
Filling missing discount fields with `0` while replacing missing region codes with `'Unassigned'`.

### AI/ML Example
Preventing data leakage by calculating imputation parameters exclusively on training sets.

import pandas as pd

# Comprehensive missing handling strategy
df_strat = df.copy()

df_strat['A'] = df_strat['A'].interpolate()  # Time/sequential interpolation
df_strat['B'] = df_strat['B'].fillna(0)       # Zero fill for counts/amounts
df_strat['C'] = df_strat['C'].fillna('Missing') # Categorical fill

print("Strategy Applied:\n", df_strat)

## Minimum 5 Interview Questions with Answers

1. **What is the difference between `dropna(how='any')` and `dropna(how='all')`?**  
   * **Answer:** `how='any'` drops a row/column if at least one entry is `NaN`. `how='all'` drops a row/column only when every single element in that slice is `NaN`.

2. **When should you use `interpolate()` instead of `fillna()`?**  
   * **Answer:** Use `interpolate()` for ordered, sequential, or time-series data where surrounding data points provide a natural estimation trend. Use `fillna()` for constant, summary statistic, or categorical fills.

3. **How do you calculate the percentage of missing values per column in a DataFrame?**  
   * **Answer:** Use `df.isnull().mean() * 100`.

4. **What is the risk of imputing missing feature values using column mean?**  
   * **Answer:** Mean imputation reduces variance artificially, distorts feature distributions, and underestimates standard error when missingness volume is high.

5. **How does `notnull()` differ from `isnull()`?**  
   * **Answer:** They are exact logical inverses. `isnull()` returns `True` for missing values, while `notnull()` returns `True` for valid non-missing elements.
