## 1. Quick Data Preview (`head`, `tail`, `sample`)

### Concept Explanation
Before running transformations, inspecting top rows (`head`), bottom rows (`tail`), and random slices (`sample`) provides an immediate reality check on raw data alignment and formatting.

### Real-world Example
Checking the top 5 rows of a climate dataset to verify sensor reading timestamps are loaded correctly.

### Business Example
Sampling random rows from an e-commerce transaction export to spot-check order formatting.

### AI/ML Example
Inspecting the tail end of a dataset to ensure no truncated rows or malformed text snippets exist at file boundaries.

```python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry'],
    'Department': ['Tech', 'Tech', 'HR', 'Marketing', 'Tech', 'Sales', 'HR', 'Tech'],
    'Salary': [75000, 85000, 50000, 60000, 90000, 55000, 52000, 88000],
    'Age': [28, 32, 25, 29, 35, 22, 27, 40]
})

print("First 3 Rows:\n", df.head(3))
print("\nLast 2 Rows:\n", df.tail(2))
print("\nRandom 2 Rows:\n", df.sample(2, random_state=42))
```

## 2. Structural Properties (`shape`, `columns`, `dtypes`)

### Concept Explanation
`shape` returns the dimensions `(rows, columns)`, `columns` lists header names as an Index object, and `dtypes` reports data types assigned to each column.

### Real-world Example
Confirming that an uploaded geographic shapefile table matches expected spatial row/column dimensions.

### Business Example
Iterating over `df.columns` to sanitize column names (e.g., removing trailing spaces or special characters).

### AI/ML Example
Checking `dtypes` to verify that numeric feature columns weren't mistakenly parsed as generic string objects (`object`).

```python
import pandas as pd

print("Shape (Rows, Columns):", df.shape)
print("\nColumn Names:", df.columns.tolist())
print("\nData Types:\n", df.dtypes)
```

## 3. Technical Metadata & Memory (`info`, `memory_usage`)

### Concept Explanation
`df.info()` prints concise summary metadata (non-null counts, data types, memory footprints), while `df.memory_usage()` inspects precise memory consumption per column in bytes.

### Real-world Example
Checking total memory usage before loading millions of satellite log records into memory.

### Business Example
Identifying memory hogging text columns in customer feedback tables to optimize DataFrame memory efficiency.

### AI/ML Example
Monitoring memory consumption during large feature matrix creation to prevent out-of-memory (OOM) crashes during model training.

```python
import pandas as pd

print("--- DataFrame Info ---")
df.info()

print("\n--- Memory Usage per Column (Bytes) ---")
print(df.memory_usage(deep=True))
```

## 4. Summary Statistics (`describe`)

### Concept Explanation
`df.describe()` generates descriptive summary statistics including count, mean, standard deviation, min/max, and percentile distributions for numeric (and optionally categorical) columns.

### Real-world Example
Calculating average daily temperatures and extreme minimum/maximum bounds across weather monitoring stations.

### Business Example
Evaluating standard deviation and median values of order totals to evaluate revenue variability.

### AI/ML Example
Detecting extreme features during Exploratory Data Analysis (EDA) that require scaling or normalization.

```python
import pandas as pd

# Summary stats for numeric columns
print("Numeric Summary Stats:\n", df.describe())

# Summary stats for non-numeric/all columns
print("\nCategorical/All Summary Stats:\n", df.describe(include='all'))
```

## 5. Unique Profiling & Frequencies (`nunique`, `value_counts`)

### Concept Explanation
`nunique()` counts total distinct values across columns, while `value_counts()` tabulates frequency distributions for unique categorical entries.

### Real-world Example
Counting unique species observed in a regional ecological field survey.

### Business Example
Analyzing distribution counts of customer payment preferences (e.g., Credit Card, UPI, PayPal).

### AI/ML Example
Checking class distributions in target labels ($y$) to detect target imbalance before training classifiers.

```python
import pandas as pd

# Unique value counts
print("Distinct Values per Column:\n", df.nunique())

# Frequency count of categories
print("\nDepartment Value Counts:\n", df['Department'].value_counts())

# Relative frequency distribution (percentages)
print("\nDepartment Proportions:\n", df['Department'].value_counts(normalize=True))
```

## Minimum 5 Interview Questions with Answers

1. **What is the difference between `df.info()` and `df.describe()`?**  
   * **Answer:** `df.info()` provides technical metadata about structural attributes (column names, non-null counts, memory usage, data types). `df.describe()` calculates statistical metrics (mean, std, min, max, percentiles) across data values.

2. **Why should you pass `deep=True` to `df.memory_usage()`?**  
   * **Answer:** By default, Pandas only measures pointer memory for object/string columns. Passing `deep=True` inspects underlying system memory consumption of object strings, returning accurate memory sizes.

3. **How does `df['col'].nunique()` differ from `len(df['col'].unique())`?**  
   * **Answer:** `nunique()` excludes `NaN` missing values by default (`dropna=True`), whereas `unique()` includes `NaN` as a distinct value in its returned array.

4. **How can `df.dtypes` reveal potential pipeline bugs after reading a CSV file?**  
   * **Answer:** If a column intended to contain numeric data (e.g., `Salary`) shows `object` in `dtypes`, it indicates dirty string data or missing value placeholders (e.g., `"$"` or `"N/A"`) that forced Pandas to cast the column as text.

5. **What parameter in `value_counts()` displays relative proportions instead of raw counts?**  
   * **Answer:** Setting `normalize=True` returns relative frequencies (percentages summing to 1.0) rather than standard absolute value counts.
