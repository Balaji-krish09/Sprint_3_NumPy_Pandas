## 1. Case Conversions & Trimming (`lower`, `upper`, `strip`)

### Concept Explanation
The `.str` accessor grants access to vectorized string methods across Pandas Series. `lower()` and `upper()` standardize text casing, while `strip()` removes leading and trailing whitespace.

### Real-world Example
Cleaning messy user inputs from web forms where names contain inconsistent capitalization and accidental spacing.

### Business Example
Standardizing store locations (e.g., `' new york '` $\rightarrow$ `'NEW YORK'`) to ensure consistent grouping across sales reports.

### AI/ML Example
Lowercasing text features in raw text datasets prior to tokenization and vectorization in NLP models.

```python
import pandas as pd

# Raw string data with whitespace and mixed cases
df = pd.DataFrame({
    'Raw_City': ['  new york ', 'LOS ANGELES  ', ' chicago ', 'HOUSTON ']
})

# Strip spaces and standardize to title/upper/lower casing
df['Clean_City'] = df['Raw_City'].str.strip().str.title()
df['Upper_City'] = df['Clean_City'].str.upper()

print("Original vs Cleaned Cities:\n", df)
```

## 2. Splitting Strings (`split`)

### Concept Explanation
`split()` breaks single string elements into lists of substrings based on a delimiter. Passing `expand=True` expands split elements directly into separate DataFrame columns.

### Real-world Example
Splitting full customer names into distinct `First_Name` and `Last_Name` columns.

### Business Example
Extracting category and subcategory labels from delimited inventory codes (e.g., `'ELEC-TV-101'`).

### AI/ML Example
Parsing complex log entries or compound feature columns into individual categorical dimensions.

```python
import pandas as pd

df_names = pd.DataFrame({'Full_Name': ['Alice Smith', 'Bob Jones', 'Charlie Brown']})

# Split full name into two separate columns using expand=True
df_names[['First_Name', 'Last_Name']] = df_names['Full_Name'].str.split(' ', expand=True)

print("Expanded Split Columns:\n", df_names)
```

## 3. String Matching & Substring Detection (`contains`)

### Concept Explanation
`contains()` tests whether a specified pattern or substring exists within each element of a Series, returning a boolean mask suitable for row filtering.

### Real-world Example
Filtering server error logs to isolate messages containing critical keyword flags like `'ERROR'` or `'FAIL'`.

### Business Example
Identifying all customer support tickets referencing specific keywords (e.g., `'refund'` or `'delay'`).

### AI/ML Example
Constructing binary indicator features based on the presence of key domain terms in text attributes.

```python
import pandas as pd

df_logs = pd.DataFrame({'Log': ['SYS_ERR_01: Memory Fail', 'SYS_OK: Boot completed', 'NET_ERR_02: Timeout']})

# Mask rows containing 'ERR'
err_mask = df_logs['Log'].str.contains('ERR')
print("Logs Containing 'ERR':\n", df_logs[err_mask])
```

## 4. Text Replacement & Regex Pattern Extraction (`replace`, Regex)

### Concept Explanation
`replace()` swaps specific characters or regular expression (regex) patterns. Setting `regex=True` allows advanced matching, extraction, and text scrubbing.

### Real-world Example
Scrubbing currency symbols, commas, or special characters from scraped pricing data before numerical casting.

### Business Example
Masking sensitive credit card digits or personal identifiable information (PII) using regular expressions.

### AI/ML Example
Removing punctuation, HTML tags, and numeric digits during text normalization pipelines.

```python
import pandas as pd

df_prices = pd.DataFrame({'Price': ['$1,200.50', '$450.00', '$99.99']})

# Remove currency symbol and commas using Regex
df_prices['Clean_Price'] = df_prices['Price'].str.replace(r'[\$,]', '', regex=True).astype(float)

print("Cleaned Numeric Prices:\n", df_prices)
```

## Minimum 5 Interview Questions with Answers

1. **Why is `expand=True` important when using `.str.split()` in Pandas?**  
   * **Answer:** By default (`expand=False`), `.str.split()` returns a Series of Python lists. Setting `expand=True` expands the split elements into a multi-column DataFrame structure, making it easy to assign to new columns.

2. **What is the difference between python string methods and Pandas `.str` accessor methods?**  
   * **Answer:** Python string methods operate on single string scalar objects. Pandas `.str` methods are vectorized functions that operate across entire Series objects while automatically handling missing (`NaN`) values without breaking.

3. **How does `contains()` handle missing values (`NaN`) by default?**  
   * **Answer:** By default, `contains()` propagates missing values, returning `NaN` for missing positions. Passing `na=False` converts missing entries to `False` boolean values for safe row filtering.

4. **How do you enable Regular Expressions inside `.str.replace()`?**  
   * **Answer:** Pass `regex=True` explicitly inside `.str.replace(pattern, replacement, regex=True)`.

5. **How can you strip non-standard trailing characters (e.g., specific punctuation) from text columns?**  
   * **Answer:** Pass the target character set to `strip()`, e.g., `df['col'].str.strip('.!?')`.
