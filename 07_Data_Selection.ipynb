## 1. Label and Index Selection (`loc`, `iloc`)

### Concept Explanation
* `loc` accesses rows and columns using explicit index labels and column names. Slicing with `loc` includes both start and stop endpoints.
* `iloc` accesses data based on integer positions (0-indexed). Slicing with `iloc` follows standard Python rules (stop index is excluded).

### Real-world Example
Retrieving patient records using explicit Patient IDs (`loc`) versus getting the first 10 entries from a raw log file regardless of labels (`iloc`).

### Business Example
Extracting metrics for specific store names (`loc`) versus isolating top 5 performing outlets by position (`iloc`).

### AI/ML Example
Selecting target feature columns by name (`loc`) versus splitting raw numerical feature matrices ($X$) by integer slicing (`iloc`).

```python
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['Tech', 'Tech', 'HR', 'Marketing', 'Tech'],
    'Salary': [75000, 85000, 50000, 60000, 90000],
    'Age': [28, 32, 25, 29, 35]
}).set_index('Name')

# loc: Label-based selection
print("--- Using loc ---")
print("Charlie's Record:\n", df.loc['Charlie'])
print("\nBob & David (Department & Salary):\n", df.loc[['Bob', 'David'], ['Department', 'Salary']])

# iloc: Position-based selection
print("\n--- Using iloc ---")
print("First 2 Rows & First 2 Columns:\n", df.iloc[0:2, 0:2])
```

## 2. Boolean Indexing & Filtering Rows (`isin`, `between`)

### Concept Explanation
Filtering selects rows matching conditions using bitwise operators (`&` for AND, `|` for OR, `~` for NOT).
* `isin()` filters rows where a column value exists in a given list.
* `between()` filters rows where numerical values fall within a specified range (inclusive by default).

### Real-world Example
Filtering weather data for days where rainfall was recorded AND temperature exceeded 30°C.

### Business Example
Extracting orders originating from specific target regions (`isin(['NY', 'CA'])`) with order amounts between $100 and $500 (`between(100, 500)`).

### AI/ML Example
Filtering out extreme outlier samples before feeding training data into a model pipeline.

```python
import pandas as pd

# Reset index to demonstrate row filtering
df_reset = df.reset_index()

# 1. Multi-condition boolean filtering (Salary > 60000 AND Dept is Tech)
tech_high_pay = df_reset[(df_reset['Salary'] > 60000) & (df_reset['Department'] == 'Tech')]

# 2. Filtering rows using isin()
target_depts = df_reset[df_reset['Department'].isin(['HR', 'Marketing'])]

# 3. Filtering rows using between()
age_25_to_30 = df_reset[df_reset['Age'].between(25, 30)]

print("Tech & Salary > 60k:\n", tech_high_pay)
print("\nIn HR or Marketing:\n", target_depts)
print("\nAge Between 25 and 30:\n", age_25_to_30)
```

## 3. The `query()` Method

### Concept Explanation
`query()` allows filtering DataFrames using concise, SQL-like string expressions. External Python variables can be referenced directly using the `@` symbol.

### Real-world Example
Filtering weather logs dynamically based on user-entered criteria without repeating `df[...]` syntax.

### Business Example
Querying inventory items matching threshold values provided by dynamic dashboard filters.

### AI/ML Example
Running dynamic dataset validation checks during automated ETL pipeline execution.

```python
import pandas as pd

min_salary = 70000
target_dept = 'Tech'

# Query string matching conditions
filtered = df_reset.query("Salary >= @min_salary and Department == @target_dept")

print("Filtered via df.query():\n", filtered)
```

## 4. Sorting Data (`sort_values`, `sort_index`)

### Concept Explanation
* `sort_values()` orders DataFrame rows based on values in one or multiple columns (ascending or descending).
* `sort_index()` orders DataFrame rows or columns based on their index labels.

### Real-world Example
Sorting athlete race completion times from fastest to slowest.

### Business Example
Sorting e-commerce products by rating score descending and price ascending.

### AI/ML Example
Sorting prediction confidence scores to display top-$K$ recommendations.

```python
import pandas as pd

# 1. Sort by single column descending
sorted_by_salary = df_reset.sort_values(by='Salary', ascending=False)

# 2. Multi-column sort (Department ascending, Salary descending)
multi_sorted = df_reset.sort_values(by=['Department', 'Salary'], ascending=[True, False])

# 3. Sort by index label
indexed_df = df.sort_index()

print("Sorted by Salary (Desc):\n", sorted_by_salary[['Name', 'Salary']])
print("\nMulti-Sort (Dept Asc, Salary Desc):\n", multi_sorted[['Name', 'Department', 'Salary']])
print("\nSorted by Index Label:\n", indexed_df)
```

## Minimum 5 Interview Questions with Answers

1. **What is the fundamental difference between `loc` and `iloc`?**  
   * **Answer:** `loc` is label-based selection using explicit row names/index labels and column names (slicing includes the endpoint). `iloc` is integer position-based selection using 0-indexed coordinates (slicing excludes the endpoint).

2. **Why must bitwise operators (`&`, `|`, `~`) be used instead of standard Python boolean operators (`and`, `or`, `not`) when filtering DataFrames?**  
   * **Answer:** Bitwise operators evaluate condition masks element-wise across Pandas Series, whereas standard Python logical operators evaluate the truth value of entire objects as a whole, causing value errors.

3. **What is the benefit of using `df.query()` over standard boolean indexing?**  
   * **Answer:** `df.query()` provides a cleaner, SQL-like string syntax that improves code readability and allows referencing environment variables seamlessly using `@`.

4. **How does `between()` handle boundaries by default?**  
   * **Answer:** By default, `between()` uses `inclusive='both'`, meaning both lower and upper boundary values are included in the result.

5. **How do you sort a DataFrame by multiple columns with different sorting orders (e.g., one Ascending, one Descending)?**  
   * **Answer:** Pass a list of column names to `by` and a corresponding list of booleans to `ascending`, such as `df.sort_values(by=['Col1', 'Col2'], ascending=[True, False])`.
