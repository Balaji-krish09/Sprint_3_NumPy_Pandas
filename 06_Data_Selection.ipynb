## 1. Inspecting Data Structure (`head`, `tail`, `sample`, `shape`)

### Concept Explanation
Before manipulating a dataset, inspecting its initial rows (`head`), ending rows (`tail`), random observations (`sample`), and dimensional footprint (`shape`) gives an immediate understanding of table layout.

### Real-world Example
Glancing at the first 5 rows of a newly downloaded government weather log to verify that headers align with data values.

### Business Example
Checking random transaction samples from a payment gateway export to audit row formatting.

### AI/ML Example
Verifying that target labels and feature columns are properly populated without shifting errors after loading a raw CSV.

```python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Department': ['HR', 'Tech', 'Tech', 'Marketing', 'Sales', 'Tech'],
    'Salary': [50000, 75000, 85000, 60000, 55000, 90000],
    'Age': [25, 30, 35, 28, 22, 40]
})

print("Shape (Rows, Columns):", df.shape)
print("\nFirst 3 Rows:\n", df.head(3))
print("\nLast 2 Rows:\n", df.tail(2))
print("\nRandom 2 Rows:\n", df.sample(2, random_state=42))
```

## 2. Summary Statistics & Metadata (`info`, `describe`)

### Concept Explanation
`df.info()` provides high-level technical metadata (data types, memory usage, non-null counts), while `df.describe()` calculates quick summary statistics (count, mean, std, min, percentiles, max) for numeric columns.

### Real-world Example
Scanning summary metrics of regional rainfall data to identify improbable recorded values.

### Business Example
Generating summary statistics across order totals to assess average order value (AOV) and detect extreme outlier purchases.

### AI/ML Example
Checking non-null counts during Exploratory Data Analysis (EDA) to locate missing feature values requiring imputation.

```python
import pandas as pd

# Inspecting DataFrame info and statistics
print("--- Technical Metadata ---")
df.info()

print("\n--- Summary Statistics ---")
print(df.describe())
```

## 3. Selecting Columns

### Concept Explanation
Columns can be selected individually returning a Series (`df['col']`), or as a subset returning a new DataFrame (`df[['col1', 'col2']]`).

### Real-world Example
Selecting only the `Temperature` column from an environmental dataset to compute daily weather metrics.

### Business Example
Extracting `Customer_ID` and `Lifetime_Value` columns to construct a dedicated CRM contact list.

### AI/ML Example
Isolating feature columns ($X$) from the target label column ($y$) during model training setup.

```python
import pandas as pd

# Selecting a single column (returns Series)
salaries = df['Salary']

# Selecting multiple columns (returns DataFrame)
tech_details = df[['Name', 'Department', 'Salary']]

print("Single Column (Series):\n", salaries.head(2))
print("\nMultiple Columns (DataFrame):\n", tech_details.head(2))
```

## 4. Label-Based Selection (`loc`)

### Concept Explanation
`df.loc[row_label, column_label]` selects data based on explicit row indices and column names. Slicing with `loc` includes both the start and the stop endpoints.

### Real-world Example
Extracting medical records for a patient identified by unique ID `'P_102'` across specific diagnostic columns.

### Business Example
Retrieving sales metrics for specific store location names across explicit financial quarter columns.

### AI/ML Example
Updating specific feature values for target rows identified by explicit sample indices.

```python
import pandas as pd

# Set Emp_ID as custom row index
df_indexed = df.set_index('Name')

# Select row by explicit label
print("David's Record:\n", df_indexed.loc['David'])

# Select specific row and column slice
print("\nCharlie's Department & Salary:\n", df_indexed.loc['Charlie', ['Department', 'Salary']])
```

---

## 5. Integer-Position Selection (`iloc`)

### Concept Explanation
`df.iloc[row_position, column_position]` selects data based on zero-based integer index positions. Standard Python slicing rules apply (stop index is exclusive).

### Real-world Example
Extracting the first 100 observations and first 3 variables from an unlabelled raw sensor matrix.

### Business Example
Grabbing the top row (highest ranking item) and last column from a pre-sorted summary table.

### AI/ML Example
Splitting tabular matrices into numerical training features without relying on column name strings ($X = df.iloc[:, :-1]$).

```python
import pandas as pd

# First 3 rows, first 2 columns using position
slice_iloc = df.iloc[0:3, 0:2]

# Last row, all columns
last_row = df.iloc[-1, :]

print("First 3 Rows & 2 Columns via iloc:\n", slice_iloc)
print("\nLast Row via iloc:\n", last_row)
```

## 6. Conditional Filtering (Boolean Indexing)

### Concept Explanation
Boolean indexing filters rows based on conditional logical statements. Multiple conditions combine using `&` (AND), `|` (OR), `~` (NOT), alongside helpers like `isin()` and `between()`.

### Real-world Example
Filtering weather records where temperature is above 30°C AND rainfall is 0mm.

### Business Example
Identifying high-value customers who spent over $1,000 OR made more than 10 purchases in a month.

### AI/ML Example
Subsetting evaluation metrics specifically for high-risk demographic groups during algorithmic bias auditing.

```python
import pandas as pd

# Filter 1: Salary > 60000 AND Department is Tech
tech_high_pay = df[(df['Salary'] > 60000) & (df['Department'] == 'Tech')]

# Filter 2: Department in explicit list
marketing_sales = df[df['Department'].isin(['Marketing', 'Sales'])]

# Filter 3: Age between 25 and 35 inclusive
age_range = df[df['Age'].between(25, 35)]

print("Tech & High Pay:\n", tech_high_pay)
print("\nAge Between 25 and 35:\n", age_range)
```

## 7. Query Method (`df.query()`)

### Concept Explanation
`df.query()` filters DataFrames using concise SQL-like string expressions, making complex multi-condition queries readable and easy to construct dynamically using `@` variables.

### Real-world Example
Filtering catalog items matching specific keyword conditions without repeating DataFrame bracket syntax.

### Business Example
Allowing dynamic threshold inputs from dashboard filters to query business tables cleanly.

### AI/ML Example
Filtering validation datasets dynamically using variable parameter thresholds during hyperparameter experiments.

```python
import pandas as pd

min_salary = 65000
target_dept = 'Tech'

# Querying using string expressions and external Python variable (@)
filtered_query = df.query("Salary >= @min_salary and Department == @target_dept")

print("Filtered via df.query():\n", filtered_query)
```
